```python
import numpy as np
import torch
import torch.nn as nn
from scipy.optimize import linear_sum_assignment
from sklearn.preprocessing import StandardScaler
import os


def iou(boxA, boxB):
    xA = max(boxA[0], boxB[0])
    yA = max(boxA[1], boxB[1])
    xB = min(boxA[0] + boxA[2], boxB[0] + boxB[2])
    yB = min(boxA[1] + boxA[3], boxB[1] + boxB[3])

    interArea = max(0, xB - xA) * max(0, yB - yA)
    if interArea == 0:
        return 0.0

    boxAArea = boxA[2] * boxA[3]
    boxBArea = boxB[2] * boxB[3]
    return interArea / float(boxAArea + boxBArea - interArea)

def centroid_distance(boxA, boxB):
    cxA = boxA[0] + boxA[2] / 2
    cyA = boxA[1] + boxA[3] / 2
    cxB = boxB[0] + boxB[2] / 2
    cyB = boxB[1] + boxB[3] / 2
    return np.sqrt((cxA - cxB) ** 2 + (cyA - cyB) ** 2)

def compute_cost(pred_box, det_box, frame_diagonal):

    dist = centroid_distance(pred_box, det_box)
    norm_dist = min(dist / (frame_diagonal + 1e-6), 1.0)
    iou_cost  = 1.0 - iou(pred_box, det_box)
    return 0.7 * norm_dist + 0.3 * iou_cost

class RichLSTMTracker(nn.Module):
    def __init__(self, input_size=12, hidden_size=128, num_layers=2):
        super(RichLSTMTracker, self).__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True, bidirectional=False)
        self.fc = nn.Sequential(
            nn.Linear(hidden_size, 64),
            nn.ReLU(),
            nn.Linear(64, 4)
        )

    def forward(self, x):
        lstm_out, _ = self.lstm(x)
        return self.fc(lstm_out[:, -1, :])

class RichGRUTracker(nn.Module):
    def __init__(self, input_size=12, hidden_size=128, num_layers=2):
        super(RichGRUTracker, self).__init__()
        self.gru = nn.GRU(input_size, hidden_size, num_layers, batch_first=True, bidirectional=False)
        self.fc = nn.Linear(hidden_size, 4)

    def forward(self, x):
        gru_out, _ = self.gru(x)
        return self.fc(gru_out[:, -1, :])


class Track:
    _id_counter = 1

    def __init__(self, initial_bbox, initial_telemetry, class_name="Unknown"):
        self.id = Track._id_counter
        Track._id_counter += 1

        self.history = []
        self.add_to_history(initial_bbox, initial_telemetry)

        self.time_since_update = 0
        self.hits = 1
        self.predicted_bbox = initial_bbox
        self.class_name = class_name

    def add_to_history(self, bbox, telemetry):
        feature_vector = [
            bbox[0], bbox[1], bbox[2], bbox[3],
            telemetry['gps_latitude'],
            telemetry['gps_longitude'],
            telemetry['altitude'],
            telemetry['gimbal_pitch'],
            telemetry['compass_heading'],
            telemetry['xspeed'],
            telemetry['yspeed'],
            telemetry['zspeed']
        ]
        self.history.append(feature_vector)
        if len(self.history) > 10:
            self.history.pop(0)

class LSTMSortTracker:
    def __init__(self, model_path=None, data_for_scaler=None, max_age=10, min_hits=3, iou_threshold=0.3):
        self.max_age = max_age
        self.min_hits = min_hits
        self.iou_threshold = iou_threshold
        self.tracks = []

        self.device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

        if model_path is not None and 'gru' in str(model_path).lower():
            self.model = RichGRUTracker().to(self.device)
        else:
            self.model = RichLSTMTracker().to(self.device)

        if model_path is not None and os.path.exists(model_path):
            self.model.load_state_dict(torch.load(model_path, map_location=self.device))
            self.model.eval()
            print(f"Loaded sequence model weights from {model_path} onto {self.device}")
        else:
            print(f"WARNING: Weights not found at {model_path}. Using random weights!")

        print("Reconstructing StandardScaler from training data...")
        raw_train = np.load(data_for_scaler)

        vels = np.zeros_like(raw_train)
        vels[:, 1:, 0] = raw_train[:, 1:, 0] - raw_train[:, :-1, 0]
        vels[:, 1:, 1] = raw_train[:, 1:, 1] - raw_train[:, :-1, 1]
        vels[:, :, 2:] = raw_train[:, :, 2:]
        vels = vels[:, 1:, :]

        self.scaler = StandardScaler()
        self.scaler.fit(vels.reshape(-1, 12))
        print("Scaler ready.")

    def predict_track(self, track, use_lstm=True):
        if not use_lstm or len(track.history) < 10:
            track.predicted_bbox = track.history[-1][:4]
            return track.predicted_bbox

        hist = np.array(track.history)
        vels = np.zeros((9, 12))
        vels[:, 0] = hist[1:, 0] - hist[:-1, 0]
        vels[:, 1] = hist[1:, 1] - hist[:-1, 1]
        vels[:, 2:] = hist[1:, 2:]

        scaled_vels = self.scaler.transform(vels)

        X_tensor = torch.tensor(scaled_vels, dtype=torch.float32).unsqueeze(0).to(self.device)
        with torch.no_grad():
            y_pred_scaled = self.model(X_tensor).cpu().numpy()[0]

        padded_pred = np.zeros((1, 12))
        padded_pred[0, :4] = y_pred_scaled
        y_pred_vel = self.scaler.inverse_transform(padded_pred)[0, :4]

        last_abs = hist[-1, :4]
        pred_bbox = [
            last_abs[0] + y_pred_vel[0],
            last_abs[1] + y_pred_vel[1],
            y_pred_vel[2],
            y_pred_vel[3]
        ]
        track.predicted_bbox = pred_bbox
        return pred_bbox

    def update(self, detections, telemetry, use_lstm=True, frame_size=(3840, 2160)):

        frame_diagonal = np.sqrt(frame_size[0] ** 2 + frame_size[1] ** 2)
        predictions = []
        for track in self.tracks:
            pred = self.predict_track(track, use_lstm=use_lstm)
            predictions.append(pred)

        matched_indices = []
        unmatched_detections = list(range(len(detections)))
        unmatched_tracks = list(range(len(self.tracks)))

        if len(self.tracks) > 0 and len(detections) > 0:
            cost_matrix = np.zeros((len(self.tracks), len(detections)))
            for t, pred_box in enumerate(predictions):
                for d, det_box in enumerate(detections):
                    cost_matrix[t, d] = compute_cost(pred_box, det_box, frame_diagonal)

            row_ind, col_ind = linear_sum_assignment(cost_matrix)

            for t, d in zip(row_ind, col_ind):
                if cost_matrix[t, d] > 0.25:
                    continue
                matched_indices.append((t, d))
                unmatched_tracks.remove(t)
                unmatched_detections.remove(d)

        for t, d in matched_indices:
            self.tracks[t].add_to_history(detections[d][:4], telemetry)
            self.tracks[t].class_name = detections[d][4]
            self.tracks[t].time_since_update = 0
            self.tracks[t].hits += 1

        for t in unmatched_tracks:
            self.tracks[t].time_since_update += 1

        for d in unmatched_detections:
            new_track = Track(detections[d][:4], telemetry, class_name=detections[d][4])
            self.tracks.append(new_track)

        self.tracks = [t for t in self.tracks if t.time_since_update <= self.max_age]

        outputs = []
        for t in self.tracks:
            if t.hits >= self.min_hits or t.time_since_update == 0:
                current_box = t.history[-1][:4]
                outputs.append([current_box[0], current_box[1], current_box[2], current_box[3], t.id, t.class_name])

        return outputs

    def update_coast(self, telemetry, use_lstm=True):

        outputs = []
        for track in self.tracks:
            pred_box = self.predict_track(track, use_lstm=use_lstm)

            track.add_to_history(pred_box, telemetry)


            if track.hits >= self.min_hits or track.time_since_update == 0:
                outputs.append([pred_box[0], pred_box[1], pred_box[2], pred_box[3], track.id, track.class_name])

        return outputs
```

```python
import cv2
import json
import zipfile
import tarfile
import os
import sys
import re
import numpy as np
from pathlib import Path
import torch
import torch.nn as nn
import timm
import torchvision.transforms as transforms
from PIL import Image
from ultralytics import YOLO
from sahi.predict import get_sliced_prediction
from sahi.models.ultralytics import UltralyticsDetectionModel


def get_video_id_from_foldername(json_path, folder_name):
    with open(json_path, 'r') as f:
        data = json.load(f)
    for img in data['images']:
        if img['source'].get('folder_name') == folder_name:
            return img['video_id']
    return None

def load_telemetry_map(json_path, target_video_id):
    print(f"Loading telemetry for video_id={target_video_id}...")
    with open(json_path, 'r') as f:
        data = json.load(f)
    telemetry_map = {}
    for img in data['images']:
        if img['video_id'] == target_video_id:
            meta = img.get('meta', {})
            telemetry_map[img['frame_index']] = {
                'gps_latitude':    meta.get('gps_latitude',    0.0),
                'gps_longitude':   meta.get('gps_longitude',   0.0),
                'altitude':        meta.get('altitude',         0.0),
                'gimbal_pitch':    meta.get('gimbal_pitch',     0.0),
                'compass_heading': meta.get('compass_heading',  0.0),
                'xspeed':          meta.get('xspeed',           0.0),
                'yspeed':          meta.get('yspeed',           0.0),
                'zspeed':          meta.get('zspeed',           0.0),
            }
    print(f"  -> {len(telemetry_map)} frames found.")
    return telemetry_map

DEFAULT_TELEMETRY = {k: 0.0 for k in
    ('gps_latitude','gps_longitude','altitude','gimbal_pitch',
     'compass_heading','xspeed','yspeed','zspeed')}

def frame_sort_key(p: Path):
    nums = re.findall(r'\d+', p.stem)
    return int(nums[-1]) if nums else 0

def extract_archive(archive_path: Path, extract_dir: Path):
    if extract_dir.exists():
        print(f"Already extracted -> {extract_dir}")
        return
    print(f"Extracting {archive_path.name} -> {extract_dir} (this may take a minute) ...")
    extract_dir.mkdir(parents=True, exist_ok=True)
    suffix = ''.join(archive_path.suffixes).lower()
    if '.zip' in suffix:
        with zipfile.ZipFile(archive_path, 'r') as zf:
            zf.extractall(extract_dir)
    elif '.tar' in suffix or '.tgz' in suffix:
        mode = 'r:gz' if suffix.endswith('.gz') or suffix.endswith('.tgz') else 'r:'
        with tarfile.open(archive_path, mode) as tf:
            tf.extractall(extract_dir, filter='data')
    else:
        raise ValueError(f"Unsupported archive format: {suffix}")
    print("Extraction complete.")

def run_demo_from_archive(archive_path: str, output_path: str = None, frames_count=500, sahi=True, model_config=None):
    if model_config is None:
        model_config = HEAVY

    yolo_path = model_config["YOLO"]
    vit_path = model_config["VIT"]
    lstm_path = model_config["SEQUENCE"]
    data_path = model_config["DATA"]

    archive_path = Path(archive_path)

    if not archive_path.exists():
        print(f"ERROR: archive not found at {archive_path}"); return

    stem = archive_path.name.replace(''.join(archive_path.suffixes), '')
    extract_dir = archive_path.parent / stem

    extract_archive(archive_path, extract_dir)

    IMAGE_EXTS = {'.png', '.jpg', '.jpeg', '.bmp'}
    images = sorted(
        [p for p in extract_dir.rglob('*') if p.suffix.lower() in IMAGE_EXTS],
        key=frame_sort_key
    )
    if not images:
        print("ERROR: No images found inside the archive."); return
    print(f"Found {len(images)} images.")

    json_path = Path('/content/instances_train_objects_in_water.json')
    folder_candidate = stem[:8]
    video_id = get_video_id_from_foldername(str(json_path), folder_candidate)
    if video_id is None:
        print(f"  WARNING: Could not find video_id for '{folder_candidate}' in JSON. "
              "Falling back to zero telemetry.")
        telemetry_map = {}
    else:
        telemetry_map = load_telemetry_map(str(json_path), video_id)

    if sahi:
        print("Loading YOLO model via SAHI ...")
        sahi_model = UltralyticsDetectionModel(
            model_path=yolo_path,
            confidence_threshold=0.65,
            device='cuda',
        )
    else:
        print("Loading standard YOLO model ...")
        standard_yolo = YOLO(yolo_path)
        standard_yolo.to('cuda')

    print("Loading classifier via timm ...")
    if model_config == HEAVY:
        classifier = timm.create_model('swin_base_patch4_window7_224', num_classes=5)
    else:
        classifier = timm.create_model('mobilevit_s', num_classes=5)

    ckpt = torch.load(vit_path, map_location='cuda')
    state_dict = ckpt['model_state_dict']
    classifier.load_state_dict(state_dict)
    classifier.to('cuda')
    classifier.eval()

    idx_to_class = ckpt['idx_to_class']

    vit_transform = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])

    print("Loading LSTM-SORT tracker …")
    tracker = LSTMSortTracker(model_path=lstm_path, data_for_scaler=data_path, max_age=10, min_hits=5)

    sample = cv2.imread(str(images[0]))
    height, width = sample.shape[:2]
    fps = 30

    if output_path is None:
        out_dir = archive_path.parent / "outputs"
        out_dir.mkdir(exist_ok=True)
        version = 1
        while (out_dir / f"output_tracked_{stem}_SPARSE_v{version}.mp4").exists():
            version += 1
        output_path = str(out_dir / f"output_tracked_{stem}_SPARSE_v{version}.mp4")

    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))
    print(f"Output -> {output_path}  ({width}x{height} @ {fps} fps)")

    np.random.seed(42)
    colors = {}

    for frame_idx, img_path in enumerate(images):
        frame = cv2.imread(str(img_path))
        if frame is None:
            print(f"  [WARN] Could not read {img_path}, skipping.")
            continue

        if frame_idx >= frames_count:
            print(f"Reached {frames_count} frames.")
            break
        if model_config == HEAVY:
          INTERVAL = 3
        else:
          INTERVAL = 1

        yolo_boxes = []
        current_telemetry = telemetry_map.get(frame_idx, DEFAULT_TELEMETRY)
        if frame_idx % INTERVAL == 0:
            frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            pil_img = Image.fromarray(frame_rgb)

            raw_detections = []
            if sahi:
                result = get_sliced_prediction(
                    pil_img,
                    sahi_model,
                    slice_height=640,
                    slice_width=640,
                    overlap_height_ratio=0.2,
                    overlap_width_ratio=0.2,
                    verbose=0,
                )
                for pred in result.object_prediction_list:
                    bbox = pred.bbox
                    raw_detections.append({
                        'x': bbox.minx, 'y': bbox.miny,
                        'w': bbox.maxx - bbox.minx, 'h': bbox.maxy - bbox.miny,
                        'cls_name': pred.category.name, 'score': pred.score.value
                    })
            else:
                results = standard_yolo(pil_img, conf=0.65, verbose=False)
                for box in results[0].boxes:
                    x1, y1, x2, y2 = box.xyxy[0].cpu().numpy()
                    raw_detections.append({
                        'x': float(x1), 'y': float(y1),
                        'w': float(x2 - x1), 'h': float(y2 - y1),
                        'cls_name': standard_yolo.names[int(box.cls[0])],
                        'score': float(box.conf[0])
                    })

            for det in raw_detections:
                x = det['x']
                y = det['y']
                w = det['w']
                h = det['h']
                cls_name = det['cls_name']
                score = det['score']

                skip_box = False
                if score < 0.90:
                    crop_x1 = max(0, int(x))
                    crop_y1 = max(0, int(y))
                    crop_x2 = min(width, int(x + w))
                    crop_y2 = min(height, int(y + h))

                    if crop_x2 > crop_x1 and crop_y2 > crop_y1:
                        crop = pil_img.crop((crop_x1, crop_y1, crop_x2, crop_y2))
                        tensor_crop = vit_transform(crop).unsqueeze(0).to('cuda')

                        with torch.no_grad():
                            vit_out = classifier(tensor_crop)
                            probs = torch.nn.functional.softmax(vit_out, dim=1)
                            vit_conf, pred_idx_tensor = torch.max(probs, dim=1)
                            vit_conf = vit_conf.item()
                            pred_idx = pred_idx_tensor.item()
                            vit_cls_name = idx_to_class[pred_idx]

                            if vit_conf < 0.50:
                                skip_box = True
                            else:
                                cls_name = f"{vit_cls_name} (ViT)"

                if not skip_box:
                    yolo_boxes.append([x, y, w, h, cls_name, score])

            active_tracks = tracker.update(yolo_boxes, current_telemetry,
                                           frame_size=(width, height))
        else:
            active_tracks = tracker.update_coast(current_telemetry)

        for t in active_tracks:
            x, y, w, h, track_id, cls_name = t
            if track_id not in colors:
                colors[track_id] = tuple(int(c) for c in
                    np.random.randint(80, 255, 3))
            color = colors[track_id]

            cv2.rectangle(frame,
                          (int(x), int(y)), (int(x+w), int(y+h)),
                          color, 4)

            label = f"ID:{track_id} | {cls_name}"
            font_scale = 0.9
            (tw, th), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, font_scale, 2)
            banner_y = max(int(y) - 35, 0)
            cv2.rectangle(frame,
                          (int(x), banner_y), (int(x) + tw + 10, banner_y + 35),
                          color, -1)
            cv2.putText(frame, label,
                        (int(x) + 5, banner_y + 25),
                        cv2.FONT_HERSHEY_SIMPLEX, font_scale, (255, 255, 255), 2)

        for tk in tracker.tracks:
            if tk.time_since_update == 0 and len(tk.history) > 1 and tk.id in colors:
                color = colors[tk.id]
                for i in range(1, len(tk.history)):
                    p, c = tk.history[i-1], tk.history[i]
                    pcx, pcy = int(p[0]+p[2]/2), int(p[1]+p[3]/2)
                    ccx, ccy = int(c[0]+c[2]/2), int(c[1]+c[3]/2)
                    cv2.line(frame, (pcx, pcy), (ccx, ccy), color, 3)

        out.write(frame)

        if frame_idx % 50 == 0:
            print(f"  Frame {frame_idx}/{len(images)}")

    out.release()
    print(f"\nDone! Saved to -> {output_path}")
```

```python
import gradio as gr
import cv2
import os
import json
import pathlib
import subprocess
import tarfile
import re

def process_video(video_id_to_find, frames_count, sahi_enabled):
    frames_count = int(frames_count)
    json_path = '/content/instances_train_objects_in_water.json'

    output_dir = '/content/videos'
    os.makedirs(output_dir, exist_ok=True)

    sahi_str = "sahi" if sahi_enabled else "yolo"
    base_vid_id = video_id_to_find.split('_id')[0] if '_id' in video_id_to_find else video_id_to_find

    original_video_name = f"{output_dir}/{video_id_to_find}_{frames_count}_original.mp4"
    archive_name = f"{output_dir}/{video_id_to_find}.tar.gz"
    compressed_orig = f"{output_dir}/{video_id_to_find}_{frames_count}_orig_comp.mp4"

    if not os.path.exists(original_video_name):
        if not os.path.exists(archive_name):
             yield None, None, None, f"Error: Archive {archive_name} does not exist."
             return

        extract_dir = pathlib.Path(archive_name).parent / video_id_to_find
        yield None, None, None, f"Extracting {archive_name}..."
        extract_archive(pathlib.Path(archive_name), extract_dir)

        search_root = pathlib.Path(extract_dir)

        IMAGE_EXTS = {'.png', '.jpg', '.jpeg', '.bmp'}
        images = sorted(
            [p for p in search_root.rglob('*') if p.suffix.lower() in IMAGE_EXTS],
            key=frame_sort_key
        )

        if not images:
            yield None, None, None, "Error: Could not locate image files on disk."
            return

        sample_img = cv2.imread(str(images[0]))
        if sample_img is None:
            yield None, None, None, "Error: Could not read sample image."
            return

        h, w, _ = sample_img.shape
        video = cv2.VideoWriter(original_video_name, cv2.VideoWriter_fourcc(*'mp4v'), 60, (w, h))

        for img_path in images[:frames_count]:
            img = cv2.imread(str(img_path))
            if img is not None: video.write(img)
        video.release()

        subprocess.run(["ffmpeg", "-i", original_video_name, "-vcodec", "libx264", "-crf", "28", compressed_orig, "-y"])

    yield compressed_orig, None, None, "Original video ready. Running HEAVY model pipeline..."

    heavy_tracked_name = f"{output_dir}/{video_id_to_find}_{frames_count}_{sahi_str}_HEAVY_tracked.mp4"
    heavy_comp_name = f"{output_dir}/{video_id_to_find}_{frames_count}_{sahi_str}_HEAVY_track_comp.mp4"

    if not os.path.exists(heavy_tracked_name) or not os.path.exists(heavy_comp_name):
        Track._id_counter = 1
        run_demo_from_archive(
            archive_path=archive_name,
            output_path=heavy_tracked_name,
            frames_count=frames_count,
            sahi=sahi_enabled,
            model_config=HEAVY
        )
        subprocess.run(["ffmpeg", "-i", heavy_tracked_name, "-vcodec", "libx264", "-crf", "28", heavy_comp_name, "-y"])

    yield compressed_orig, heavy_comp_name, None, "HEAVY model complete. Running LITE model pipeline..."

    lite_tracked_name = f"{output_dir}/{video_id_to_find}_{frames_count}_{sahi_str}_LITE_tracked.mp4"
    lite_comp_name = f"{output_dir}/{video_id_to_find}_{frames_count}_{sahi_str}_LITE_track_comp.mp4"

    if not os.path.exists(lite_tracked_name) or not os.path.exists(lite_comp_name):
        Track._id_counter = 1
        run_demo_from_archive(
            archive_path=archive_name,
            output_path=lite_tracked_name,
            frames_count=frames_count,
            sahi=sahi_enabled,
            model_config=LITE
        )
        subprocess.run(["ffmpeg", "-i", lite_tracked_name, "-vcodec", "libx264", "-crf", "28", lite_comp_name, "-y"])

    yield compressed_orig, heavy_comp_name, lite_comp_name, "Processing Complete!"


available_videos = []
output_dir = '/content/videos'
if os.path.exists(output_dir):
    for f in os.listdir(output_dir):
        if f.endswith('.tar.gz') and '_id' in f:
            base_name = f.replace('.tar.gz', '')
            if base_name not in available_videos:
                available_videos.append(base_name)
available_videos.sort()

if not available_videos:
    available_videos = ["DJI_0069_id9_images", "DJI_0057_id0_images"]

with gr.Blocks() as demo:
    gr.Markdown("# 🚁 SeaDronesSee Object Tracking")
    gr.Markdown("Select a Video ID to reconstruct the frames, run the tracker pipeline, and view the results.")

    with gr.Row():
        with gr.Column(scale=1):
            vid_id = gr.Dropdown(choices=available_videos, label="Video Name", value=available_videos[0])
            frames = gr.Number(label="Frames to Process", value=200, minimum=10, maximum=5000)
            sahi = gr.Checkbox(label="Use SAHI (Sliced Inference)", value=True)
            btn = gr.Button("Run Pipeline", variant="primary")
            status = gr.Textbox(label="Status", interactive=False)

        with gr.Column(scale=2):
            orig_vid = gr.Video(label="Original Video")
            with gr.Row():
                heavy_vid = gr.Video(label="HEAVY Model Tracked Output")
                lite_vid = gr.Video(label="LITE Model Tracked Output")

    btn.click(fn=process_video, inputs=[vid_id, frames, sahi], outputs=[orig_vid, heavy_vid, lite_vid, status])

demo.launch(debug=True, share=True)
```
